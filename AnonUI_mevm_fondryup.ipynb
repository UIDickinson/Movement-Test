#Movement contract deployment with Foundryup and Fractal

Prerequisite: Rust, Git and Foundryup

#First setup

```python
#Set Git and curl
sudo apt-get update
sudo apt-get install git
sudo apt-get install curl

#Install Foundryup with cargo
curl -L https://foundry.paradigm.xyz | bash

#source shell
source ~/.bashrc

#verify
foundryup
forge --version

# Create a project directory and navigate into it
mkdir foundry-move-evm
cd foundry-move-evm

#initiate foundry project
forge init

#Create an .env file and add your private key:
nano .env

#Copy below and replace `<your private key>` with your private key without "0x":
PRIVATE_KEY=<your private key>
```

#Write your Solidity contract
1. Under src/ open the file Counter.sol. This is the Smart Contract that you will be deploying to M1. You can replace the contents of this file with your own contract.

```python
// SPDX-License-Identifier: UNLICENSED
pragma solidity ^0.8.13;
contract Counter {
    uint256 public number;
    function setNumber(uint256 newNumber) public {
        number = newNumber;
    }
    function increment() public {
        number++;
    }
}
```

2. Now, under script, open the file Counter.s.sol. This is a simple script that I will use to deploy our contract on M1. You can replace the contents of this file with your own script.

```python
// SPDX-License-Identifier: UNLICENSED
pragma solidity ^0.8.13;
import {Script} from "forge-std/Script.sol";
import {Counter} from "../src/Counter.sol";
contract CounterScript is Script {
    Counter public counter;
    uint256 public deployerPrivateKey = vm.envUint("PRIVATE_KEY");
    function run() public {
        vm.startBroadcast(deployerPrivateKey);
        counter = new Counter();
        vm.stopBroadcast();
    }
}
```

#Deploy your solidity contract

```python
forge script CounterScript --broadcast --chain-id 30732 --rpc-url https://mevm.devnet.imola.movementlabs.xyz --legacy
```

#Interact with your contract

Under script, create a new file Interact.s.sol and add the following contents

```python
//replace <counterAddress> with your deployed contract Address
// SPDX-License-Identifier: UNLICENSED
pragma solidity ^0.8.13;
import {Script} from "forge-std/Script.sol";
import {Counter} from "../src/Counter.sol";
import "forge-std/console.sol";
contract InteractScript is Script {
    Counter public counter;
    uint256 public deployerPrivateKey = vm.envUint("PRIVATE_KEY");
    function run() public {
        vm.startBroadcast(deployerPrivateKey);
        counter = Counter(<counterAddress>);
        counter.setNumber(1);
        console.log("Number is now: ", counter.number());
        counter.increment();
        console.log("Number was incremented to: ", counter.number());
        vm.stopBroadcast();
    }
}
```

```python
#Run this command below
forge script InteractScript --broadcast --chain-id 30732 --rpc-url https://mevm.devnet.imola.movementlabs.xyz
```
